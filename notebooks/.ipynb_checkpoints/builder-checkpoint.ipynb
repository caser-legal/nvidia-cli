# 🚀 iOS App Builder
### Powered by NVIDIA Nemotron-3-Nano

## Prerequisites

Get your NVIDIA NIM API Key: [Click here](https://build.nvidia.com/nvidia/nemotron-3-nano-30b-a3b?signin=true&api_key=true)

````python
import getpass, os, subprocess, re, json
from IPython.display import display, Markdown

key = getpass.getpass("Enter NVIDIA API key: ")
assert key.startswith("nvapi-"), "Invalid key"
os.environ["NVIDIA_API_KEY"] = key

print("✅ API Key configured")
````

````python
from openai import OpenAI

client = OpenAI(base_url="https://integrate.api.nvidia.com/v1", api_key=os.environ["NVIDIA_API_KEY"])
MODEL = "nvidia/nemotron-3-nano-30b-a3b"

def ask(prompt, system="You are an expert iOS/SwiftUI developer."):
    r = client.chat.completions.create(model=MODEL, messages=[
        {"role": "system", "content": system},
        {"role": "user", "content": prompt}
    ], temperature=0.3, max_tokens=4096)
    return r.choices[0].message.content

def bash(cmd, timeout=120):
    r = subprocess.run(cmd, shell=True, capture_output=True, text=True, timeout=timeout)
    return (r.stdout + r.stderr).strip()

def write_file(name, content):
    with open(os.path.join(PROJECT_PATH, name), 'w') as f: f.write(content)
    print(f"✅ {name}")

def show(code): display(Markdown(f"```swift\n{code}\n```"))

print("✅ Nemotron-3-Nano ready")
````

---
# Phase 1: Describe Your App

````python
APP_DESC = input("📱 Describe your app idea: ")
````

````python
# LLM suggests app name
suggestion = ask(f"Suggest a short, catchy app name (one word, PascalCase) for: {APP_DESC}. Reply with ONLY the name.")
suggested_name = re.sub(r'[^a-zA-Z0-9]', '', suggestion.strip().split()[0])
print(f"🤖 Suggested name: {suggested_name}")
````

````python
APP_NAME = input(f"📱 App Name [{suggested_name}]: ") or suggested_name
PROJECT_PATH = f"/Users/home/Documents/iOS/{APP_NAME}/{APP_NAME}"
XCODEPROJ = f"/Users/home/Documents/iOS/{APP_NAME}/{APP_NAME}.xcodeproj"
print(f"✅ {APP_NAME}")
````

````python
if not os.path.exists(PROJECT_PATH):
    print(f"⚠️ Create Xcode project first:")
    print(f"1. Xcode → File → New → Project → App")
    print(f"2. Product Name: {APP_NAME}")
    print(f"3. Save to: /Users/home/Documents/iOS/{APP_NAME}/")
    input("Press Enter when done...")
print(f"✅ Project ready at {PROJECT_PATH}")
````

---
# Phase 2: Data Model

````python
model_prompt = f"""For an iOS app: {APP_DESC}

Design a Swift data model. Reply with ONLY a JSON object:
{{"name": "ModelName", "properties": [["propName", "Type"], ...]}}

Types: String, Int, Bool, Date, Double. Keep it simple (3-5 properties)."""

model_json = ask(model_prompt)
match = re.search(r'\{[^{}]*\}', model_json, re.DOTALL)
if match:
    model_data = json.loads(match.group())
    MODEL = model_data["name"]
    PROPS = [tuple(p) for p in model_data["properties"]]
else:
    MODEL, PROPS = "Item", [("name", "String")]

print(f"🤖 LLM suggests: {MODEL}")
for n, t in PROPS: print(f"   • {n}: {t}")
````

````python
accept = input("Accept model? [y]: ") or "y"
if accept.lower() == "n":
    MODEL = input("Model name: ") or MODEL
    print("Enter properties (name: Type), empty when done:")
    PROPS = []
    while True:
        p = input()
        if not p: break
        if ":" in p:
            n, t = [x.strip() for x in p.split(":", 1)]
            if t in ["String", "Int", "Bool", "Date", "Double"]: PROPS.append((n, t))
    if not PROPS: PROPS = [("name", "String")]

DISPLAY_PROP = next((n for n,t in PROPS if t=="String"), PROPS[0][0])
DISPLAY_TYPE = next((t for n,t in PROPS if n==DISPLAY_PROP), "String")
print(f"✅ {MODEL}: {', '.join(f'{n}:{t}' for n,t in PROPS)}")
````

---
# Phase 3: Features

````python
features_prompt = f"""For app: {APP_DESC} with model {MODEL}
Recommend features. Reply ONLY with JSON:
{{"nav": "list" or "tabs", "tabs": ["Tab1", "Tab2"] if tabs, "edit": true/false, "search": true/false}}"""

feat_json = ask(features_prompt)
match = re.search(r'\{[^{}]*\}', feat_json, re.DOTALL)
if match:
    feat = json.loads(match.group())
    NAV_STYLE = "tabs" if feat.get("nav") == "tabs" else "list"
    TABS = feat.get("tabs", [])
    EDIT = feat.get("edit", True)
    SEARCH = feat.get("search", False) and DISPLAY_TYPE == "String"
else:
    NAV_STYLE, TABS, EDIT, SEARCH = "list", [], True, False

print(f"🤖 LLM suggests:")
print(f"   Nav: {NAV_STYLE}" + (f" ({', '.join(TABS)})" if TABS else ""))
print(f"   Edit: {EDIT}, Search: {SEARCH}")
````

````python
accept = input("Accept features? [y]: ") or "y"
if accept.lower() == "n":
    NAV_STYLE = input("Nav style (list/tabs) [list]: ") or "list"
    if NAV_STYLE == "tabs":
        TABS = input("Tab names (comma-separated): ").split(",")
        TABS = [t.strip() for t in TABS if t.strip()]
    EDIT = input("Include edit? [y]: ").lower() != "n"
    SEARCH = DISPLAY_TYPE == "String" and input("Include search? [n]: ").lower() == "y"
print(f"✅ Features configured")
````

---
# Phase 4: Generate Swift Code

````python
props_str = ", ".join(f"{n}: {t}" for n,t in PROPS)
code = ask(f"""Write {MODEL}.swift: struct {MODEL}: Identifiable, Codable, Hashable with let id: UUID and properties: {props_str}. Include init. Reply with ONLY Swift code.""")
code = re.sub(r'^```swift\n?|```$', '', code.strip(), flags=re.MULTILINE).strip()
write_file(f"{MODEL}.swift", code)
show(code)
````

````python
code = ask(f"""Write {MODEL}Manager.swift: @Observable class with var items: [{MODEL}] = [], CRUD methods (add, update, delete, delete at IndexSet), find(id: UUID) -> {MODEL}?, persist to UserDefaults key "{APP_NAME}Data". Reply with ONLY Swift code.""")
code = re.sub(r'^```swift\n?|```$', '', code.strip(), flags=re.MULTILINE).strip()
write_file(f"{MODEL}Manager.swift", code)
show(code)
````

````python
code = ask(f"""Write {MODEL}Row.swift: struct {MODEL}Row: View with let item: {MODEL}, display first 2 properties ({', '.join(f'{n}:{t}' for n,t in PROPS[:2])}) in HStack. Reply with ONLY Swift code.""")
code = re.sub(r'^```swift\n?|```$', '', code.strip(), flags=re.MULTILINE).strip()
write_file(f"{MODEL}Row.swift", code)
show(code)
````

````python
code = ask(f"""Write {MODEL}FormView.swift: @Bindable var manager: {MODEL}Manager, var item: {MODEL}? for add/edit, @State vars for {props_str}, Form in Section, NavigationStack, toolbar Cancel/Save, onAppear load from item. Reply with ONLY Swift code.""")
code = re.sub(r'^```swift\n?|```$', '', code.strip(), flags=re.MULTILINE).strip()
write_file(f"{MODEL}FormView.swift", code)
show(code)
````

````python
code = ask(f"""Write {MODEL}DetailView.swift: @Bindable var manager, let itemID: UUID, computed var current: {MODEL} using manager.find(id:) with fallback, List with LabeledContent for {props_str}, Delete button, {'Edit toolbar button with sheet to FormView' if EDIT else 'no edit'}. Reply with ONLY Swift code.""")
code = re.sub(r'^```swift\n?|```$', '', code.strip(), flags=re.MULTILINE).strip()
write_file(f"{MODEL}DetailView.swift", code)
show(code)
````

````python
search_desc = f", @State search with filtered computed on {DISPLAY_PROP}, .searchable" if SEARCH else ""
code = ask(f"""Write {MODEL}ListView.swift: @Bindable var manager, @State showingAdd{search_desc}, ContentUnavailableView when empty, List ForEach NavigationLink(value: item.id), .onDelete, .navigationDestination(for: UUID.self) to DetailView, toolbar + button sheet for add, title "{APP_NAME}". Reply with ONLY Swift code.""")
code = re.sub(r'^```swift\n?|```$', '', code.strip(), flags=re.MULTILINE).strip()
write_file(f"{MODEL}ListView.swift", code)
show(code)
````

````python
if NAV_STYLE == "tabs" and TABS:
    prompt = f"""Write ContentView.swift: @State var manager = {MODEL}Manager(), TabView with tabs {', '.join(TABS)}, first tab NavigationStack with {MODEL}ListView, others placeholder Text, each tab in NavigationStack. Add #Preview. Reply with ONLY Swift code."""
else:
    prompt = f"""Write ContentView.swift: @State var manager = {MODEL}Manager(), NavigationStack with {MODEL}ListView(manager: manager). Add #Preview. Reply with ONLY Swift code."""
code = ask(prompt)
code = re.sub(r'^```swift\n?|```$', '', code.strip(), flags=re.MULTILINE).strip()
write_file("ContentView.swift", code)
show(code)
````

---
# Phase 5: Build & Run

````python
FILES = [f"{MODEL}.swift", f"{MODEL}Manager.swift", f"{MODEL}Row.swift",
         f"{MODEL}FormView.swift", f"{MODEL}DetailView.swift", f"{MODEL}ListView.swift", "ContentView.swift"]
print("📁 Generated:", ", ".join(FILES))
print(f"\n⚠️ Add files to Xcode:")
print(f"1. Open {XCODEPROJ}")
print(f"2. Right-click {APP_NAME} folder → Add Files")
print(f"3. Select all .swift files (not {APP_NAME}App.swift)")
print("4. Uncheck 'Copy items if needed' → Add")
input("\nPress Enter when done...")
````

````python
print("🔨 Building...")
out = bash(f"xcodebuild -project '{XCODEPROJ}' -scheme '{APP_NAME}' -destination 'generic/platform=iOS' build 2>&1", timeout=300)
if "BUILD SUCCEEDED" in out:
    print("✅ BUILD SUCCEEDED")
else:
    print("❌ BUILD FAILED")
    for line in out.split("\n"):
        if "error:" in line.lower(): print(f"  {line[:120]}")
````
