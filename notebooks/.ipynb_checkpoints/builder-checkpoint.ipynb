# 🚀 iOS App Builder
### Powered by NVIDIA Nemotron 3 Nano (1M Context)

## Prerequisites

Get your NVIDIA NIM API Key: [build.nvidia.com](https://build.nvidia.com)

````python
import getpass, os, subprocess, re, json
from IPython.display import display, Markdown

# Check for existing key in environment
key = os.environ.get("NVIDIA_API_KEY") or os.environ.get("NGC_API_KEY")
if not key:
    key = getpass.getpass("Enter NVIDIA API key: ")
assert key.startswith("nvapi-"), "Invalid key - must start with nvapi-"
os.environ["NVIDIA_API_KEY"] = key
os.environ["NGC_API_KEY"] = key
print("✅ API Key configured")
````

````python
from openai import OpenAI
import time

client = OpenAI(base_url="https://integrate.api.nvidia.com/v1", api_key=os.environ["NVIDIA_API_KEY"])
MODEL = "nvidia/nemotron-3-nano-30b-a3b"  # 1M context, MoE architecture

def ask(prompt, system="You are an expert iOS/SwiftUI developer. Write clean, modern Swift code.", retries=3, max_tokens=4096):
    for i in range(retries):
        try:
            r = client.chat.completions.create(model=MODEL, messages=[
                {"role": "system", "content": system},
                {"role": "user", "content": prompt}
            ], temperature=0.3, max_tokens=max_tokens)
            return r.choices[0].message.content
        except Exception as e:
            if i < retries - 1:
                wait = 2 ** (i + 1)  # Exponential backoff
                print(f"⚠️ Retry {i+1}/{retries} in {wait}s: {e}")
                time.sleep(wait)
            else:
                raise

def bash(cmd, timeout=300):
    r = subprocess.run(cmd, shell=True, capture_output=True, text=True, timeout=timeout)
    return (r.stdout + r.stderr).strip()

def write_file(name, content):
    path = os.path.join(PROJECT_PATH, name)
    os.makedirs(os.path.dirname(path) if '/' in name else PROJECT_PATH, exist_ok=True)
    with open(path, 'w') as f: f.write(content)
    print(f"✅ {name}")

def extract_code(text):
    m = re.search(r'```(?:swift)?\n([\s\S]*?)```', text)
    return m.group(1).strip() if m else text.strip()

# Device config
DEVICE_ID = "00008110-00061CD921A3A01E"  # iPhone 14

print(f"✅ {MODEL} ready")
````

---
# Phase 1: Describe Your App

````python
APP_DESC = input("📱 Describe your app idea: ")
````

````python
suggestion = ask(f"Suggest a short, catchy app name (one word, PascalCase) for: {APP_DESC}. Reply with ONLY the name, nothing else.")
suggested_name = re.sub(r'[^a-zA-Z0-9]', '', suggestion.strip().split()[0])
print(f"🤖 Suggested: {suggested_name}")
````

````python
APP_NAME = input(f"📱 App Name [{suggested_name}]: ") or suggested_name
PROJECT_PATH = f"~/Documents/iOS/{APP_NAME}/{APP_NAME}"
XCODEPROJ = f"~/Documents/iOS/{APP_NAME}/{APP_NAME}.xcodeproj"
print(f"✅ {APP_NAME}")
````

````python
if not os.path.exists(PROJECT_PATH):
    print(f"⚠️ Create Xcode project first:")
    print(f"1. Xcode → File → New → Project → App")
    print(f"2. Product Name: {APP_NAME}")
    print(f"3. Interface: SwiftUI, Language: Swift")
    print(f"4. Save to: ~/Documents/iOS/")
    input("Press Enter when done...")
print(f"✅ Project ready at {PROJECT_PATH}")
````

---
# Phase 2: Data Model

````python
model_prompt = f"""For an iOS app: {APP_DESC}

Design a Swift data model. Reply with ONLY a JSON object:
{{"name": "ModelName", "properties": [["propName", "Type"], ...]}}

Types: String, Int, Bool, Date, Double, [String], UUID. Keep it focused (3-6 properties)."""

model_json = ask(model_prompt)
match = re.search(r'\{[^{}]*\}', model_json, re.DOTALL)
if match:
    model_data = json.loads(match.group())
    MODEL_NAME = model_data["name"]
    PROPS = [tuple(p) for p in model_data["properties"]]
else:
    MODEL_NAME, PROPS = "Item", [("name", "String")]

print(f"🤖 Model: {MODEL_NAME}")
for n, t in PROPS: print(f"   • {n}: {t}")
````

````python
accept = input("Accept model? [y]: ") or "y"
if accept.lower() == "n":
    MODEL_NAME = input("Model name: ") or MODEL_NAME
    print("Enter properties (name: Type), empty when done:")
    PROPS = []
    while True:
        p = input()
        if not p: break
        if ":" in p:
            n, t = [x.strip() for x in p.split(":", 1)]
            PROPS.append((n, t))
    if not PROPS: PROPS = [("name", "String")]

DISPLAY_PROP = next((n for n,t in PROPS if t=="String"), PROPS[0][0])
print(f"✅ {MODEL_NAME} with display property: {DISPLAY_PROP}")
````

---
# Phase 3: Features

````python
features_prompt = f"""For app: {APP_DESC} with model {MODEL_NAME}
Recommend features. Reply ONLY with JSON:
{{"nav": "list" or "tabs", "tabs": ["Tab1", "Tab2"] if tabs, "edit": true/false, "search": true/false, "swipeActions": true/false}}"""

feat_json = ask(features_prompt)
match = re.search(r'\{[^{}]*\}', feat_json, re.DOTALL)
if match:
    feat = json.loads(match.group())
    NAV_STYLE = "tabs" if feat.get("nav") == "tabs" else "list"
    TABS = feat.get("tabs", [])
    EDIT = feat.get("edit", True)
    SEARCH = feat.get("search", False)
    SWIPE = feat.get("swipeActions", True)
else:
    NAV_STYLE, TABS, EDIT, SEARCH, SWIPE = "list", [], True, False, True

print(f"🤖 Nav: {NAV_STYLE}" + (f" ({', '.join(TABS)})" if TABS else ""))
print(f"   Edit: {EDIT}, Search: {SEARCH}, Swipe: {SWIPE}")
````

````python
accept = input("Accept features? [y]: ") or "y"
if accept.lower() == "n":
    NAV_STYLE = input("Nav style (list/tabs): ") or NAV_STYLE
    if NAV_STYLE == "tabs":
        TABS = [t.strip() for t in input("Tab names (comma-separated): ").split(",")]
    EDIT = input("Edit feature? (y/n): ").lower() != "n"
    SEARCH = input("Search feature? (y/n): ").lower() == "y"
    SWIPE = input("Swipe actions? (y/n): ").lower() != "n"
print(f"✅ Features confirmed")
````

---
# Phase 4: Generate Swift Files

````python
# Generate Model with SwiftData
props_str = ', '.join(f'{n}: {t}' for n,t in PROPS)
model_code = ask(f"""Write a Swift model for {MODEL_NAME} with properties: {props_str}

Requirements:
- Use @Model macro (SwiftData, iOS 17+)
- Make it Identifiable with UUID id
- Include a static sample property for previews
- Add Codable conformance

Reply with ONLY the Swift code, no explanation.""")
model_code = extract_code(model_code)
write_file(f"{MODEL_NAME}.swift", model_code)
display(Markdown(f"```swift\n{model_code}\n```"))
````

````python
# Generate Row View
row_code = ask(f"""Write a SwiftUI row view for {MODEL_NAME} showing {DISPLAY_PROP}.

Requirements:
- Clean, modern design with proper spacing
- Use SF Symbols where appropriate
- Support Dynamic Type
- Show 2-3 key properties in the row

Reply with ONLY the Swift code.""")
row_code = extract_code(row_code)
write_file(f"{MODEL_NAME}Row.swift", row_code)
display(Markdown(f"```swift\n{row_code}\n```"))
````

````python
# Generate Detail View
detail_code = ask(f"""Write a SwiftUI detail view for {MODEL_NAME}.

Requirements:
- Show all properties: {props_str}
- Clean layout with sections
- Navigation title
- Edit button in toolbar if editable

Reply with ONLY the Swift code.""")
detail_code = extract_code(detail_code)
write_file(f"{MODEL_NAME}DetailView.swift", detail_code)
display(Markdown(f"```swift\n{detail_code}\n```"))
````

````python
# Generate Edit View (if enabled)
if EDIT:
    edit_code = ask(f"""Write a SwiftUI edit/add form for {MODEL_NAME}.

Requirements:
- Properties: {props_str}
- Use @Environment(\.dismiss) for dismissal
- Form with proper sections
- Save and Cancel buttons in toolbar
- Validate required fields
- Use @Bindable for the model (SwiftData)

Reply with ONLY the Swift code.""")
    edit_code = extract_code(edit_code)
    write_file(f"{MODEL_NAME}EditView.swift", edit_code)
    display(Markdown(f"```swift\n{edit_code}\n```"))
else:
    print("⏭️ Edit view skipped")
````

````python
# Generate List View
search_str = f"Include searchable modifier filtering by {DISPLAY_PROP}" if SEARCH else "No search needed"
swipe_str = "Add swipe actions for delete" if SWIPE else "No swipe actions"
list_code = ask(f"""Write a SwiftUI list view for {MODEL_NAME}.

Requirements:
- Use @Query for SwiftData
- NavigationStack with proper title
- Show {MODEL_NAME}Row for each item
- {search_str}
- {swipe_str}
- Add button (+) in toolbar to create new items
- NavigationLink to detail view
- Empty state when no items

Reply with ONLY the Swift code.""")
list_code = extract_code(list_code)
write_file(f"{MODEL_NAME}ListView.swift", list_code)
display(Markdown(f"```swift\n{list_code}\n```"))
````

````python
# Generate ContentView
if NAV_STYLE == "tabs" and TABS:
    tabs_str = ', '.join(TABS)
    content_code = ask(f"""Write SwiftUI ContentView with TabView.

Requirements:
- Tabs: {tabs_str}
- First tab shows {MODEL_NAME}ListView
- Use SF Symbols for tab icons
- Modern iOS 17+ style

Reply with ONLY the Swift code.""")
else:
    content_code = ask(f"""Write SwiftUI ContentView that shows {MODEL_NAME}ListView.

Requirements:
- Simple wrapper view
- iOS 17+ style

Reply with ONLY the Swift code.""")
content_code = extract_code(content_code)
write_file("ContentView.swift", content_code)
display(Markdown(f"```swift\n{content_code}\n```"))
````

````python
# Generate App Entry with SwiftData
app_code = ask(f"""Write the @main App struct for {APP_NAME}App.

Requirements:
- Use SwiftData modelContainer for {MODEL_NAME}
- WindowGroup with ContentView
- iOS 17+ deployment target

Reply with ONLY the Swift code.""")
app_code = extract_code(app_code)
write_file(f"{APP_NAME}App.swift", app_code)
display(Markdown(f"```swift\n{app_code}\n```"))
````

---
# Phase 5: Build & Deploy

````python
print("🔨 Building for device...")
build_cmd = f'xcodebuild -project "{XCODEPROJ}" -scheme "{APP_NAME}" -destination "id={DEVICE_ID}" build 2>&1'
result = bash(build_cmd, timeout=300)

if "BUILD SUCCEEDED" in result:
    print("✅ BUILD SUCCEEDED")
    BUILD_SUCCESS = True
else:
    # Show last 30 lines for errors
    lines = result.split('\n')
    print('\n'.join(lines[-30:]))
    print("\n⚠️ Build failed - check errors above")
    BUILD_SUCCESS = False
````

````python
# Auto-fix common build errors
if not BUILD_SUCCESS:
    print("🔧 Attempting auto-fix...")
    
    # Extract error messages
    errors = [l for l in result.split('\n') if 'error:' in l.lower()]
    if errors:
        fix_prompt = f"""These Swift build errors occurred:
{chr(10).join(errors[:5])}

Suggest fixes. Reply with JSON: {{"file": "filename.swift", "issue": "description", "fix": "what to change"}}"""
        
        fix_suggestion = ask(fix_prompt)
        print(f"🤖 Suggestion: {fix_suggestion}")
        
        retry = input("\nRetry build after manual fixes? [y]: ") or "y"
        if retry.lower() == "y":
            result = bash(build_cmd, timeout=300)
            if "BUILD SUCCEEDED" in result:
                print("✅ BUILD SUCCEEDED")
                BUILD_SUCCESS = True
````

````python
# Deploy to device
if BUILD_SUCCESS:
    print("📱 Installing to device...")
    
    # Find the built app
    app_path = bash(f'find ~/Library/Developer/Xcode/DerivedData/{APP_NAME}-*/Build/Products/Debug-iphoneos -name "*.app" -type d 2>/dev/null | head -1')
    
    if app_path:
        print(f"Found: {app_path}")
        install_result = bash(f'ios-deploy --bundle "{app_path}" --id {DEVICE_ID}', timeout=120)
        
        if "Install" in install_result and "100%" in install_result:
            print(f"✅ {APP_NAME} installed on device!")
        else:
            print(install_result)
            print("\n⚠️ Install may have failed - check device")
    else:
        print("⚠️ Could not find built app - try building in Xcode")
else:
    print("⏭️ Skipping deploy - build failed")
````
