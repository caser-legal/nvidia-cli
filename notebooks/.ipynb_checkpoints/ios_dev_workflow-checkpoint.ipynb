## 📦 Setup: Core Tools

````python
import subprocess, os, json, datetime
from pathlib import Path
from IPython.display import display, Markdown, HTML, Image

NVIDIA_CLI_PATH = "/Users/home/Documents/nvidia-cli"
PROJECT_PATH = "/Users/home/Documents/iOS/YourApp/YourApp"  # CHANGE THIS

def bash(cmd, cwd=None, timeout=120):
    try:
        r = subprocess.run(cmd, shell=True, cwd=cwd or PROJECT_PATH, capture_output=True, text=True, timeout=timeout,
            env={**os.environ, "PATH": f"/opt/homebrew/bin:/usr/local/bin:{os.environ.get('PATH', '')}" })
        return (r.stdout + (f"\n[stderr] {r.stderr}" if r.stderr else "")) or "[Done]"
    except Exception as e: return f"[Error] {e}"

def file_read(path):
    p = path if path.startswith("/") else os.path.join(PROJECT_PATH, path)
    try:
        with open(p) as f: return f.read()
    except Exception as e: return f"[Error] {e}"

def file_write(path, content):
    p = path if path.startswith("/") else os.path.join(PROJECT_PATH, path)
    try:
        os.makedirs(os.path.dirname(p), exist_ok=True)
        with open(p, 'w') as f: f.write(content)
        return f"✅ Wrote {len(content)} chars to {path}"
    except Exception as e: return f"[Error] {e}"

def list_files(path=".", pattern="*"):
    p = path if path.startswith("/") else os.path.join(PROJECT_PATH, path)
    try:
        return "\n".join([f"{'📁' if i.is_dir() else '📄'} {i.name}" for i in sorted(Path(p).glob(pattern))]) or "[Empty]"
    except Exception as e: return f"[Error] {e}"

def show_swift(code): display(Markdown(f"```swift\n{code}\n```"))
def screenshot():
    p = f"/tmp/sim_{datetime.datetime.now().strftime('%H%M%S')}.png"
    bash(f"xcrun simctl io booted screenshot '{p}'")
    display(Image(filename=p, width=300)); return p

print(f"✅ Tools loaded | 📁 {PROJECT_PATH}")
````
